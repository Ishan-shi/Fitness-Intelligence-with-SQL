```python
fitness_cols = [
    'Age',
    'Weight (kg)',
    'Max_BPM',
    'Avg_BPM',
    'Resting_BPM',
    'Session_Duration (hours)',
    'Calories_Burned',
    'Fat_Percentage',
    'Water_Intake (liters)',
    'BMI',
    'Carbs',
    'Proteins',
    'Fats',
    'Calories',
    'lean_mass_kg'
]

corr = df1[fitness_cols].corr()

plt.figure(figsize=(14,10))

sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Fitness Intelligence Correlation Heatmap')

plt.show()
```

```python
sns.scatterplot(data=df1,x='Session_Duration (hours)',y='Calories_Burned',hue='Workout_Type')
plt.title("Workout Duration vs Calories Burned")
plt.show()
```

```python
sns.regplot(data=df1,x='Avg_BPM',y='Calories_Burned')
plt.title('Avg_BPM vs Calories Burned')
plt.show()
```

```python
sns.scatterplot(data=df1,x='protein_per_kg',y='lean_mass_kg',hue='Gender')
plt.title("Protein Vs Lean Mass")
plt.show()
```

```python
workout_burn= df1.groupby('Workout_Type')['Calories_Burned'].mean().sort_values(ascending=False)

sns.barplot(x=workout_burn,y= workout_burn.index)
plt.title("Average Calories Burned by Workout Type")
plt.show()
```

```python
muscle_burn= df1.groupby('Target Muscle Group')['Burns Calories (per 30 min)'].mean().sort_values(ascending= False)
plt.figure(figsize=(12,6))
sns.barplot(x=muscle_burn,y=muscle_burn.index)
plt.title('Average Calories burnt by Muscle Group')
plt.xlabel('Calories Burned')
plt.ylabel('Muscle Group')
plt.show()
```

```python
pivot_eq = pd.pivot_table(
    df1,
    values='Burns Calories (per 30 min)',
    index='Equipment Needed',
    columns='Difficulty Level',
    aggfunc='mean'
)

plt.figure(figsize=(14,8))

sns.heatmap(
    pivot_eq,
    annot=True,
    cmap='coolwarm'
)

plt.title('Equipment vs Difficulty Level')

plt.show()
```
